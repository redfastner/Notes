# Tokenization

Converts raw human text into a sequence of discrete numerical IDs.

Algorithms used: Byte-Pair Encoding (BPE), WordPiece, or Unigram Language Model.

It uses a learned frequency-based lookup dictionary to greedily merge common character pairs (or raw bytes) into subwords.

$$\text{"Understanding"} \rightarrow [\text{"Under"}, \text{"standing"}] \rightarrow [1420, 8831]$$

# Dense Embedding Lookup

Converts discrete integer IDs into continuous, high-dimensional vector representations.

Performs an $O(1)$ lookup mapping token ID $i$ to row $i$ in an embedding matrix of size $(\text{Vocabulary Size} \times d_{\text{model}})$.$$x_i = \text{Lookup}(ID_i)$$

# Positional Encoding / Injecting Sequence Order

Injects spatial sequence order into the token vectors (since matrix math alone is permutation-invariant).

Algorithms used:

- Sinusoidal Positional Encoding (Original Transformer): Uses fixed sine/cosine wave frequencies added directly to the embeddings.

- Rotary Position Embedding (RoPE) (Modern LLMs): Applies a rotation matrix to vector pairs in $2\text{D}$ subspaces within the attention heads during key/query computation.
